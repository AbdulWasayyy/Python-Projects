This section imports all required modules from Selenium.

webdriver controls the browser
By helps locate elements on the page
Keys allows keyboard actions like pressing Enter
NoSuchElementException is used to handle errors when elements are missing
time is used to add delays so the page can load properly

# 🔐 Step 2: Store User Credentials

Here, you define your LinkedIn login details and phone number.
These values are later used to log in and fill forms automatically.

# 🌐 Step 3: Launch Browser

The script starts a Chrome browser using Selenium and maximizes the window.
Maximizing helps ensure all elements are visible and interactable.

# 🔎 Step 4: Open LinkedIn Job Page

The bot navigates to a LinkedIn job search page with filters applied (e.g., location and job role).

# 🔑 Step 5: Sign In Process
Finds and clicks the “Sign in” button
Waits for the login page to load
Enters email and password
Submits the login form by pressing Enter

# 📩 Step 6: Click Apply Button

After logging in, the bot waits for the page to load and clicks the Easy Apply button on a job listing.

# 📱 Step 7: Fill Phone Number (if needed)

The script checks if a phone number field exists:

If present and empty → fills it
If not found → skips without crashing (using exception handling)

# 🧾 Step 9: Confirmation & Exit
Prints a success message
Waits a few seconds so you can see the result
Closes the browser

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
from selenium.common.exceptions import NoSuchElementException
import time
```

```python
ACCOUNT_EMAIL = "YOUR_LOGIN_EMAIL"
ACCOUNT_PASSWORD = "YOUR_LOGIN_PASSWORD"
PHONE = "YOUR_PHONE_NUMBER"
```

```python
driver = webdriver.Chrome()  
driver.maximize_window()
```

```python
driver.get("https://www.linkedin.com/jobs/search/?f_LF=f_AL&geoId=102257491&keywords=marketing%20intern&location=London%2C%20England%2C%20United%20Kingdom&redirect=false&position=1&pageNum=0")
```

```python
sign_in_button = driver.find_element(By.LINK_TEXT, "Sign in")
sign_in_button.click()
```

```python
time.sleep(5)
email_field = driver.find_element(By.ID, "username")
email_field.send_keys(ACCOUNT_EMAIL)

password_field = driver.find_element(By.ID, "password")
password_field.send_keys(ACCOUNT_PASSWORD)
password_field.send_keys(Keys.ENTER)
```

```python
time.sleep(5)
apply_button = driver.find_element(By.CSS_SELECTOR, ".jobs-s-apply button")
apply_button.click()

time.sleep(5)
try:
    phone = driver.find_element(By.CLASS_NAME, "fb-single-line-text__input")
    if phone.get_attribute("value") == "":   
        phone.send_keys(PHONE)
except NoSuchElementException:
    print("No phone field found, skipping...")

submit_button = driver.find_element(By.CSS_SELECTOR, "footer button")
submit_button.click()
```
